```python
import google.generativeai as genai
```

```python
from google.colab import userdata
```

```python
#CONFIFGURE THE API KEY IN GOOGLE AI STUDIO
genai.configure(
    api_key=userdata.get('GOOGLE_API_KEY')
)
```

```python
print("Available models")

for m in genai.list_models():
    print(f"{m.name} ({m.supported_generation_methods})")
```

```python
#load gemini model
model = genai.GenerativeModel('gemini-2.5-flash-lite') # gemini-2.0-flash-lite might be charged less #/gemini-pro-latest  - best model
```

```python
#define a very simple function
```

```python
def simple_agent(prompt):
  response = model.generate_content(prompt)
  return response.text

```

```python
#my first prompt
question = 'explain ai agent'
answer = simple_agent(question)

print('ques:', question)
print('ans:',answer)
```
