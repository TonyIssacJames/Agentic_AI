# What is RAG?
- RAG stands for Retrieval-Augmented Generation. It is a technique in natural language processing that combines retrieval-based methods with generative models to improve the quality and relevance of generated text. RAG allows models to access external knowledge sources, such as databases or documents, to enhance their responses and provide more accurate and informative outputs.- In RAG, the model retrieves relevant information from a knowledge base or document collection based on the input query. This retrieved information is then used to generate a response that is more contextually relevant and informative. RAG can be particularly useful in tasks like question answering, where the model needs to access specific information to provide accurate answers. By leveraging retrieval techniques, RAG can enhance the capabilities of generative models and improve their performance on various natural language processing tasks.
- RAG is a powerful approach that combines the strengths of both retrieval-based and generative models, allowing for more accurate and contextually relevant responses in natural language processing tasks.
- RAG can be implemented using various techniques, such as using a retriever model to fetch relevant documents or passages from a knowledge base, and then using a generator model to produce a response based on the retrieved information. This approach allows for more efficient and effective generation of text, as the model can leverage external knowledge to enhance its responses.
- RAG has been shown to be effective in various applications, including question answering, dialogue systems, and text generation tasks. By combining retrieval and generation, RAG can provide more accurate and informative responses, making it a valuable technique in natural language processing.
- Overall, RAG is a powerful technique that combines retrieval-based methods with generative models to enhance the quality and relevance of generated text. It allows models to access external knowledge sources, improving their performance on various natural language processing tasks and providing more accurate and informative responses.

## A simplified RAG workflow:
1. **Input Query**: The user provides an input query or prompt that requires a response
2. **Retrieval**: The RAG model retrieves relevant information from a knowledge base or document collection based on the input query. This can involve using a retriever model to identify relevant documents or passages that contain information related to the query.
3. **Generation**: The retrieved information is then used to generate a response. A generator model takes the retrieved information and the input query to produce a response that is contextually relevant and informative.
4. **Output**: The generated response is returned to the user as the final output.

```mermaid
flowchart TD
    A[Input Query] --> B[Retrieval]
    C[Document Corpus] --> B
    B --> D[Retrieved Context]
    D --> E[LLM]
    A --> E
    E --> F[Generatied Answer]
```

## Type of chunking strategies:
1. **Fixed-size chunking**: This strategy involves dividing the text into fixed-size chunks, such as sentences, paragraphs, or a specific number of tokens. For example, if we have a long document, we can split it into sentences or paragraphs to create smaller chunks of text that can be processed by the retriever model.
2. **Semantic chunking**: This strategy involves dividing the text based on its semantic content. For example, we can use natural language processing techniques to identify topics or themes in the text and create chunks based on those topics. This can help ensure that the retrieved information is more relevant to the input query.
3. **Overlap chunking**: This strategy involves creating overlapping chunks of text to ensure that important information is not missed during retrieval. For example, if we have a long document, we can create chunks that overlap by a certain number of tokens or sentences to ensure that the retriever model captures all relevant information.    
4. **Hierarchical chunking**: This strategy involves creating a hierarchy of chunks, where larger chunks are divided into smaller sub-chunks. For example, we can first divide a document into sections, and then further divide each section into paragraphs or sentences. This allows for more granular retrieval of information based on the input query.

**Example of chunking strategy**:

```python
text = "The capital of France is Paris. It is known for its iconic landmarks such as the Eiffel Tower and the Louvre Museum."
# Fixed-size chunking
sentences = text.split('. ')
print(sentences)
# Output: ['The capital of France is Paris', 'It is known for its iconic landmarks such as the Eiffel Tower and the Louvre Museum'] 
# Semantic chunking : 
topics = ["capital", "landmarks"]
chunks = []
for topic in topics:
    if topic in text:
        chunks.append(text)
print(chunks)
# Output: ['The capital of France is Paris. It is known for its iconic landmarks such as the Eiffel Tower and the Louvre Museum.']
# Overlap chunking
overlap_chunks = []
for i in range(0, len(text), 20):       
    overlap_chunks.append(text[i:i+40])  # Create overlapping chunks of 40 characters
print(overlap_chunks)
# Output: ['The capital of France is Paris. It is kn', 's known for its iconic landmarks such as the Eiff', 'el Tower and the Louvre Museum.']
# Hierarchical chunking
sections = text.split('. ')
hierarchical_chunks = []
for section in sections:
    hierarchical_chunks.append(section)
print(hierarchical_chunks)
# Output: ['The capital of France is Paris', 'It is known for its iconic landmarks such as the Eiffel Tower and the Louvre Museum']
```

## Why RAG Systems need Evaluation?
1. A RAG system can fail in more ways that a promp-only llm. You need to evaliate both the retriever and generator components of the RAG system to ensure that they are working effectively together. 
2. The retriever component may fail to retrieve relevant information, while the generator component may fail to generate accurate and informative responses based on the retrieved information. 
3. Evaluating both components is crucial to identify and address any issues in the RAG system and improve its overall performance.

Failure | What happens | Example Sympom
--- | --- | ---
Hallucination | Model produces false information | The Answers citea a policy that does not exist in the retrieved document.
Poor retrieval | Retrieves misses important chunks | Answer is vague and lacks specific details that are present in the retrieved documents.
Irrelevant context | Retrieves return noisy chunks | Answer is off-topic and does not address the input query effectively.
Weak answer | The Moedel receive useful context but answer poorly | Answer is incomplete or too generic, failing to leverage the retrieved information effectively.
Missing ground truth | No correct answer in the retrieved context | Answer is incorrect or misleading because the necessary information was not retrieved.
